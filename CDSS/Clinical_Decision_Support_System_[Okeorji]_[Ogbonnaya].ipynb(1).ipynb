```python
%python

def check_medication_safety(
    patient_allergies,
    prescribed_med,
    prescribed_treatment=None
):
    """
    Rule-Based Medication Safety Checker using if/else logic
    This is NOT AI - it follows predefined rules
    """
    alerts = []
    # Rule 1: Check peanut allergy
    if "peanut" in patient_allergies and "peanut" in prescribed_med.lower():
        alerts.append("Peanut Allergy detected")
        alerts.append("Recommendation: Use Epinephrine instead")

    # Rule 2: Check egg allergy
    if "egg" in patient_allergies and "egg" in prescribed_med.lower():
        alerts.append("Egg Allergy detected")
        alerts.append("Recommendation: Use loratadine instead")

    # Rule 3: Check for food allergy medications
    if "food allergy" in patient_allergies and "food allergy" in prescribed_med.lower():
        alerts.append("Food allergy detected")
        alerts.append("Recommendation: Use antihistamines instead")

    return alerts

# Test the rule-based system
patient = {
    "name": "Ben Tennyson",
    "allergies": ["peanuts", "tree nuts"],
    "age": 10
}

medication = "Epinephrine"
safety_check = check_medication_safety(
    patient["allergies"],
    medication
)

print(f"Patient: {patient['name']}")
print(f"Prescribed: {medication}")
print(f"\nSafety Check Results:")
if safety_check:
    for alert in safety_check:
        print(alert)
        injection = "epinephrine"
        print(f"Recommendation: Use {injection} instead")
else:
    print("Treatment, emergency.")


patient = {
    "name": "Wayne Brown",
    "allergies": ["egg", "wheat"],
    "age": 24
}

medication = "loratadine"
safety_check = check_medication_safety(
    patient["allergies"],
    medication
)

print(f"Patient: {patient['name']}")
print(f"Prescribed: {medication}")
print(f"\nSafety Check Results:")
if safety_check:
    for alert in safety_check:
        print(alert)
        oral = "loratadine"
        print(f"Recommendation: Use {oral} instead")
else:
    print("No safety alerts detected.")
```
